```python
import pandas as pd
import numpy as np
```

```python
data = {
    "customer_id": [101,102,103,104,105,106,107,108,109,110],
    "customer_name": ["Ram", "Sita", "Hari", None, "Gita", "Gita", "Mina", "Ravi", None, "Anita"],
    "age": [25, 30, np.nan, 28, 35, 35, 22, np.nan, 40, 27],
    "city": ["Kathmandu", "Pokhara", "Kathmandu", "Butwal", "Pokhara", "Pokhara", "Butwal", "Kathmandu", "Pokhara", "Kathmandu"],
    "purchase_amount": [1500, 2000, 1800, np.nan, 3000, 3000, 1200, 2500, 4000, 1700],
    "payment_method": ["Cash", "Card", "Cash", "Online", "Card", "Card", "Cash", "Online", "Card", "Cash"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print(df[df.duplicated(keep=False)])
```

```python
print(df.loc[4:5])
```

```python
df.loc[5, "customer_id"] = 105

print("Duplicate count:", df.duplicated().sum())
```

```python
df= df.drop_duplicates()
print("Duplicate count after cleaning:", df.duplicated().sum())
print("Updated dataset shape:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["customer_name"] = df["customer_name"].fillna("Unknown")

print(df["customer_name"])
```

```python
median_age = df["age"].median()

df["age"] = df["age"].fillna(median_age)

print("Median Age:", median_age)
print(df["age"])
```

```python
median_purchase = df["purchase_amount"].median()

df["purchase_amount"] = df["purchase_amount"].fillna(median_purchase)

print("Median Purchase Amount:", median_purchase)
print(df["purchase_amount"])
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

```python
print(df)
print("\nDataset Shape:", df.shape)
```

```python
df.to_csv("cleaned_customers.csv", index=False)

print("Cleaned dataset saved successfully!")
```
