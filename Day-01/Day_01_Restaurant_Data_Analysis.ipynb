```python
#import pandas

import pandas as pd
import numpy as np
```

```python
#load the dataset

df= pd.read_csv("tips.csv")
```

```python
#inspect the dataset

print(df.head(10))
print(df.tail(5))
print(df.shape)
print(df.columns)
df.info()
print(df.describe())
```

```python
#check data quality

df.isnull().sum()
print(df.duplicated().sum())
```

```python
#calculate business metrices

total_bill = df["total_bill"].sum()

average_bill = df["total_bill"].mean()

average_tip = df["tip"].mean()

maximum_party = df["size"].max()

print("Total Bill:", total_bill)
print("Average Bill:", average_bill)
print("Average Tip:", average_tip)
print("Maximum Party Size:", maximum_party)
```

```python
#create a new business metrices

df["tip_percentage"] = (
    df["tip"] / df["total_bill"]
) * 100

print(df[["total_bill", "tip",
          "tip_percentage"]].head())
```

```python
smoker_analysis = df.groupby("smoker").agg(
    average_tip=("tip", "mean"),
    average_tip_percentage=("tip_percentage", "mean")
)

print(smoker_analysis)
```

```python
#Which day generates the highest total bill?

df.groupby("day")["total_bill"].sum()
```

```python
#Which day has the highest average tip?

df.groupby("day")["tip"].mean()
```

```python
#Compare lunch and dinner customers

df.groupby("time")["total_bill"].mean()
```

```python
#Compare smokers and non-smokers
df.groupby("smoker")["tip"].mean()
```

```python
assert df.shape[0] > 0
assert df.shape[1] > 0

assert "total_bill" in df.columns
assert "tip" in df.columns

assert "tip_percentage" in df.columns

assert df["tip_percentage"].notna().all()

print("Day 1 basic verification passed!")
```

```python
#Analyze daily revenue and customers

daily_analysis = df.groupby("day").agg(
    total_revenue=("total_bill", "sum"),
    average_bill=("total_bill", "mean"),
    total_customers=("size", "sum")
)

print(daily_analysis)
```
